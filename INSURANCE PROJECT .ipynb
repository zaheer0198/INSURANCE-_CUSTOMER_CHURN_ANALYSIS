**REMOVE DUPLICATES**

```python
df.duplicated().sum()
```

```python
duplicates = df[df.duplicated()]
duplicates.head()
```

```python
print('duplicate rows:',df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
df.isnull().sum()
```

FOR NUMERICAL DATE

```python
df["annual_income"] = df["annual_income"].fillna(df["annual_income"].median())

df["customer_satisfaction_score"] = df[
    "customer_satisfaction_score"
    ].fillna(df["customer_satisfaction_score"].median())
```

FOR CATEGORICAL COLUMNS

```python
df["occupation"] = df["occupation"].fillna(
    df["occupation"].mode()[0]
)

df["payment_method"] = df["payment_method"].fillna(
    df["payment_method"].mode()[0]
)
```

FOR DATES


```python
df["last_contact_date"] = pd.to_datetime(
    df["last_contact_date"],
    errors="coerce"
)
```

```python
df.isnull().sum()
```

```python
df["policy_start_date"] = pd.to_datetime(
    df["policy_start_date"],
    errors="coerce")

df["last_contact_date"] = pd.to_datetime(
    df["last_contact_date"],
    errors="coerce")

df["renewal_due_date"] = pd.to_datetime(
    df["renewal_due_date"],
    errors="coerce")
```

```python
df['renewal_status'].value_counts()
```

```python
analysis_df = df[
    df["renewal_status"].isin(["Renewed", "Churned"])].copy()
```

```python
analysis_df.shape
```

```python
analysis_df["churn_flag"] = np.where(
    analysis_df["renewal_status"] == "Churned",1,0)
```

```python
analysis_df["churn_flag"].value_counts()
```

```python
churn_rate = analysis_df["churn_flag"].mean() * 100

print("Customer Churn Rate:", round(churn_rate, 2), "%")
```

RENEWAL V/S CHURN

```python
status_count = analysis_df["renewal_status"].value_counts()

print(status_count)
```

```python
plt.figure(figsize=(4, 2))

sns.countplot(
    data=analysis_df,
    x="renewal_status"
)

plt.title("Customer Renewal vs Churn")
plt.xlabel("Renewal Status")
plt.ylabel("Number of Customers")

plt.show()
```

Renewal and churn percentage

```python
churn_percentage = (
    analysis_df["renewal_status"]
    .value_counts(normalize=True)
    .mul(100)
)

print(churn_percentage)
```

```python
plt.figure(figsize=(7, 5))

plt.pie(
    churn_percentage,
    labels=churn_percentage.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Customer Renewal vs Churn Percentage")

plt.show()
```

CUSTOMERS SATISFACTION

```python
satisfaction_churn = (
    analysis_df
    .groupby("customer_satisfaction_score")["churn_flag"]
    .mean()
    .mul(100)
)

print(satisfaction_churn)
```

ANALYZE SUPPORT CALL

ANALYZE CLAIMS

```python
claim_churn = (
    analysis_df
    .groupby("claims_count")["churn_flag"]
    .mean()
    .mul(100)
)

print(claim_churn)
```

ANALYZE PREMIUM

```python
analysis_df["premium_group"] = pd.qcut(
    analysis_df["annual_premium"],
    q=4,
    labels=[
        "Low Premium",
        "Medium-Low Premium",
        "Medium-High Premium",
        "High Premium"
    ]
)
```

```python
premium_churn = (
    analysis_df
    .groupby("premium_group", observed=True)["churn_flag"]
    .mean()
    .mul(100)
)

print(premium_churn)
```

```python
plt.figure(figsize=(4, 2))

sns.barplot(
    x=premium_churn.index,
    y=premium_churn.values
).axes.set_facecolor("k")

plt.title("Churn Rate by Premium Group")
plt.xlabel("Premium Group")
plt.ylabel("Churn Rate (%)")

plt.xticks(rotation=30)
plt.show()
```

ANALYZE INCOME


```python
analysis_df["income_group"] = pd.qcut(
    analysis_df["annual_income"],
    q=4,
    labels=[
        "Low Income",
        "Lower-Middle Income",
        "Upper-Middle Income",
        "High Income"
    ]
)
```

```python
income_churn = (
    analysis_df
    .groupby("income_group", observed=True)["churn_flag"]
    .mean()
    .mul(100)
)

print(income_churn)
```

ANALYZE AGE

```python
analysis_df['age_group'] = pd.cut(
    analysis_df['age'],
    bins=[17, 25, 35, 45, 55, 65, 100],
    labels=[
        '18-25',
        '26-30',
        '31-40',
        '41-55',
        '56-65',
        '66+'
    ]
)
```

```python
from pandas.core.groupby import groupby
age_churn = (
    analysis_df
    .groupby('age_group', observed = True)['churn_flag']
    .mean()
    .mul(100)
)

print(age_churn)
```

```python
plt.figure(figsize=(4,2))

sns.barplot(
    x=age_churn.index,
    y=age_churn.values
  )

plt.title('CHARN RATE BY AGE GROUP')
plt.xlabel('AGE GROUP')
plt.ylabel('CHARN RATE(%)')

plt.show()
```

ANALYZE CITY

```python
city_churn = (
    analysis_df
    .groupby("city")["churn_flag"]
    .agg(["count", "mean"])
)

city_churn["churn_rate"] = city_churn["mean"] * 100

city_churn = city_churn.sort_values(
    "churn_rate",
    ascending=False
)

print(city_churn)
```

```python
plt.figure(figsize=(6, 2))

sns.barplot(
    data=city_churn.reset_index(),
    x="city",
    y="churn_rate"
)

plt.title("Churn Rate by City")
plt.xlabel("City")
plt.ylabel("Churn Rate (%)")

plt.xticks(rotation=45)

plt.show()
```

**CORRELATION MATRIX**

```python
numeric_columns = [
    "age",
    "annual_income",
    "annual_premium",
    "sum_assured",
    "policy_tenure_months",
    "claims_count",
    "total_claim_amount",
    "customer_satisfaction_score",
    "support_calls",
    "late_payments",
    "churn_flag"
]

correlation = analysis_df[numeric_columns].corr()

correlation["churn_flag"].sort_values(
    ascending=False
)
```

```python
plt.figure(figsize=(8, 4))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix - Insurance Customer Churn")

plt.show()
```

```python
correlation["churn_flag"].sort_values(
    ascending=False
)
```

```python
analysis_df["risk_score"] = (
    (analysis_df["customer_satisfaction_score"] <= 4).astype(int)
    + (analysis_df["late_payments"] >= 2).astype(int)
    + (analysis_df["support_calls"] >= 5).astype(int)
    + (analysis_df["claims_count"] >= 3).astype(int)
)


analysis_df["risk_category"] = pd.cut(
    analysis_df["risk_score"],
    bins=[-1, 1, 2, 4],
    labels=["Low Risk", "Medium Risk", "High Risk"]
)

analysis_df["risk_category"].value_counts()
```

```python
risk_churn = (
    analysis_df
    .groupby("risk_category", observed=True)["churn_flag"]
    .mean()
    .mul(100)
)

print(risk_churn)
```

```python
plt.figure(figsize=(4, 2))

sns.barplot(
    x=risk_churn.index,
    y=risk_churn.values
)

plt.title("Churn Rate by Customer Risk Category")
plt.xlabel("Risk Category")
plt.ylabel("Churn Rate (%)")

plt.show()
```
